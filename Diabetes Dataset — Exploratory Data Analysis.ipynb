## 10. Diabetes Outcome Distribution

The `Outcome` variable represents the diabetes result.

We will examine how many observations belong to each outcome category.

```python
df["Outcome"].value_counts()
```

```python
df["Outcome"].value_counts(normalize = True)*100
```

```python
plt.figure(figsize=(7,6))
sns.countplot(data = df,
              x = "Outcome")
plt.title("Diabetes Outcome Distribution")
plt.xlabel("Outcome")
plt.ylabel("Number Of Patients")

plt.show()
```

### 🔎 Insight

The dataset contains two outcome categories:

- **0 → Non-diabetic**
- **1 → Diabetic**

The distribution is imbalanced, with the non-diabetic group representing a larger proportion of the observations than the diabetic group.

This imbalance should be considered if the dataset is later used for machine learning.

## 11. Zero Value Analysis

Although the dataset contains no explicit null values, certain variables may contain zero values that could represent missing or biologically unusual measurements.

We will investigate zero values in selected medical variables.

```python
zero_columns = ["Glucose","BloodPressure","SkinThickness","Insulin","BMI"]
zero_counts = (df[zero_columns]==0).sum()
zero_counts
```

```python
plt.figure(figsize = (10,5))
zero_counts.sort_values(ascending = False).plot(kind = "bar")

plt.title("Zero Values in Selected Variables")
plt.xlabel("Variables")
plt.ylabel("Number Of Zero Values")

plt.xticks(rotation = 45)

plt.show()
```

### 🔎 Insight

Several medical variables contain zero values.

For variables such as **Glucose, BloodPressure, and BMI**, a value of zero may not represent a realistic physiological measurement and may indicate unavailable data.

Therefore, these values should be investigated carefully before statistical modeling or machine learning.

## 13. Outlier Analysis

Boxplots are useful for identifying potential outliers.

An outlier is an observation that lies unusually far from the rest of the distribution.

We will visualize the numerical variables using boxplots.

```python
plt.figure(figsize = (15,8))
sns.boxplot(data = df)
plt.title("Boxplot of Numerical Variables")
plt.xticks(rotation = 45)
plt.show()
```

### 🔎 Insight

The boxplots indicate potential outliers in several variables.

The most noticeable extreme values appear in variables such as:

- Insulin
- SkinThickness
- DiabetesPedigreeFunction
- Pregnancies

These values should not automatically be removed. They should first be investigated to determine whether they represent valid observations or data-quality issues.

## 14. Correlation Analysis

Correlation helps us understand the strength and direction of linear relationships between numerical variables.

The correlation coefficient ranges from:

- **-1** → Strong negative relationship
- **0** → No linear relationship
- **+1** → Strong positive relationship

```python
correlation = df.corr()
correlation
```

```python
plt.figure(figsize = (10,8))
sns.heatmap(correlation,annot = True,
            cmap = "coolwarm",fmt="2f")
plt.title("Correlation Matrix")
plt.show()
```

### 🔎 Insight

The correlation matrix helps identify variables that have stronger relationships with the diabetes outcome.

Glucose generally shows one of the strongest positive relationships with `Outcome`.

However, correlation does not imply causation, so these relationships should be interpreted as associations rather than causal effects.

## 15. Glucose Level vs Diabetes Outcome

Glucose is an important variable to investigate because differences in glucose levels may be associated with diabetes outcomes.

We will compare glucose distributions across the two outcome groups.

```python
plt.figure(figsize =(8,5))
sns.boxplot(data = df,
            x = "Outcome",
            y = "Glucose")
plt.title("Glucose Level by Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Glucose")
plt.show()
```

### 🔎 Insight

The glucose distribution differs between the two outcome groups.

The diabetic group generally shows higher glucose values compared with the non-diabetic group.

This indicates that glucose has a noticeable association with the observed diabetes outcome in this dataset.

## 16. BMI vs Diabetes Outcome

BMI is another important health-related variable.

We will compare BMI distributions between the two outcome groups.

```python
plt.figure(figsize = (8,5))
sns.boxplot(data = df,
            x = "Outcome",
            y = "BMI")
plt.title("BMI by Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("BMI")
plt.show()
```

### 🔎 Insight

The BMI distributions show differences between the two outcome groups.

The diabetic group generally has a higher BMI distribution than the non-diabetic group, although considerable overlap exists between the groups.

## 17. Age vs Diabetes Outcome

Age can be analyzed to understand whether the distribution of age differs between the outcome groups.

```python
plt.figure(figsize = (8,5))
sns.boxplot(data = df,
            x = "Outcome",
            y = "Age")
plt.title("Age By Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Age")
plt.show()
```

### 🔎 Insight

The age distribution differs between the two outcome groups.

The diabetic group tends to contain older observations compared with the non-diabetic group, although the distributions overlap.

## 18. Pregnancies vs Diabetes Outcome

Let's investigate whether the number of pregnancies differs between the two outcome groups.

```python
plt.figure(figsize=(8,5))
sns.boxplot(data = df,
            x = "Outcome",
            y = "Pregnancies")
plt.title("Pregnancies by Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Pregnancies")
plt.show()
```

## 19. Glucose Distribution by Outcome

A violin plot combines information about distribution shape and density.

This allows us to compare glucose levels across diabetes outcome groups in more detail.

```python
plt.figure(figsize = (8,5))
sns.violinplot(data = df,
            x = "Outcome",
            y = "Glucose")
plt.title("Glucose by Distribution by Outcome")
plt.show()
```

### 🔎 Insight

The glucose distribution for the diabetic group is shifted toward higher values compared with the non-diabetic group.

This reinforces the relationship observed in the previous glucose analysis.

## 20. Multivariate Relationship Analysis

A pairplot allows us to visually examine relationships between multiple numerical variables while separating observations by diabetes outcome.

```python
sns.pairplot(df,vars = ["Glucose","BMI","Age","Pregnancies"],
             hue = "Outcome")
plt.show()
```

### 🔎 Insight

The pairplot provides a multivariate view of the dataset.

Glucose shows noticeable separation between the two outcome groups, while variables such as BMI and Age show greater overlap.

This suggests that diabetes outcome is likely associated with multiple variables rather than a single feature.

## 21. Outcome Percentage Analysis

Let's calculate the percentage of observations belonging to each diabetes outcome category.

```python
outcome_percentage = (df["Outcome"].value_counts(normalize = True).mul(100).round(2))
outcome_percentage
```

```python
plt.figure(figsize=(7, 5))

plt.pie(
    outcome_percentage,
    labels=['Non-Diabetic', 'Diabetic'],
    autopct='%1.1f%%',
    startangle=90)

plt.title('Diabetes Outcome Percentage')

plt.show()
```

### 🔎 Insight

The dataset contains a higher proportion of non-diabetic observations than diabetic observations.

This confirms that the target variable is not perfectly balanced.

## 22. Group-wise Statistical Analysis

We will calculate the average values of important numerical variables for each diabetes outcome.

This helps us compare the two groups quantitatively.

```python
group_analysis = df.groupby('Outcome')[["Pregnancies","Glucose","BloodPressure","BMI","Age"]].mean()

group_analysis
```

### 🔎 Insight

Group-wise averages provide a direct comparison between non-diabetic and diabetic observations.

The comparison can reveal variables where the average values differ noticeably between the two groups.

Glucose, BMI, Age, and Pregnancies show useful differences that can be explored further.

## 23. Features Most Associated With Outcome

To identify which numerical variables have the strongest linear association with the diabetes outcome, we sort the correlation coefficients with `Outcome`.

```python
outcome_corr = (df.corr()['Outcome'].sort_values(ascending=False))

outcome_corr
```

### 🔎 Insight

This analysis identifies which variables have stronger positive or negative linear associations with diabetes outcome.

Glucose is among the variables with the strongest positive association with `Outcome`.

These correlations can help guide further analysis, but they should not be interpreted as proof of causation.
